# Caption your photo

Open this notebook from the repository root and select the environment described in [inference/README.md](inference/README.md). Download `final_model_inference.zip` from the [model release](https://github.com/AskatQwerty/AMLNN/releases/tag/final-model-2026-11-16) and extract it under `local_artifacts/final_model/`.

Put a photo at `photo.jpg` in the repository root, or change only `IMAGE_PATH` below. Run the code cell with **Shift+Enter**. It displays the photo, caption and generation time.

The selected model uses **ResNet50 + Faster R-CNN + Transformer**, epoch 11, beam size 5. It extracts features only for this photo in memory. No training, COCO dataset or stored feature arrays are needed.

**Generation time** includes the photo's feature extraction and caption generation. Weight loading is shown separately. Each run reloads the weights. The displayed photo is resized only for the preview; the model reads the original file.

This is a clean user-facing example. Earlier private-photo outputs were removed. This portable copy was syntax-checked, but inference was not rerun during publication. Joint fine-tuning checkpoints are separate and cannot replace this caption-only model package.


```python
import sys
from pathlib import Path
from PIL import Image,ImageOps
from IPython.display import display,Markdown

PROJECT_DIR=Path.cwd()
sys.path.insert(0,str(PROJECT_DIR/"inference"))

from caption_image import caption_image

# Change only this path
IMAGE_PATH=PROJECT_DIR/"photo.jpg"

# Load the saved model and describe this photo
MODEL_DIR=PROJECT_DIR/"local_artifacts/final_model/selected_model"
ENCODER_CACHE=PROJECT_DIR/"local_artifacts/final_model/encoder_cache"
result=caption_image(IMAGE_PATH,MODEL_DIR,device="auto",torch_cache=ENCODER_CACHE)

# Show a smaller preview without changing the original image
with Image.open(IMAGE_PATH) as image:
    preview=ImageOps.exif_transpose(image).convert("RGB")

preview.thumbnail((800,600))
display(preview)
display(Markdown("**Caption:** "+result["caption"]))

# Generation includes this photo's CNN features and beam search
print(f"Generation (image features + caption): {result['inference_seconds']:.2f} seconds")
print(f"Loading saved weights: {result['loading_seconds']:.2f} seconds")
print(f"Device: {result['gpu'] or result['device']}")
```
